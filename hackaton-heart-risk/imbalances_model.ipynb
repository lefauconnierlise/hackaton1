## Imbalanced dataset

### Modify class distributions

- Oversampling : insert copies (or variations) of the positive class into the data set
- Undersampling : remove samples from the majority class


```python
def minority_class_rate(df,
                        target_name='TARGET',
                        minority_class_target_value=1):
    df_minority_class = df[df[target_name]==minority_class_target_value]
    return len(df_minority_class)/len(df)

def oversampling(df,
                 minority_class_rate=0.5,
                 target_name='TARGET',
                 minority_class_target_value=1):

    df_minority_class = df[df[target_name]==minority_class_target_value]
    n_samples_to_add = math.ceil((minority_class_rate * len(df) - len(df_minority_class))/(1-minority_class_rate))
    df_oversampled = df_minority_class.sample(n=n_samples_to_add, replace=True)

    return pd.concat([df, df_oversampled], ignore_index=True)
```

```python
print(f'Initial proportion of heart issues among samples {minority_class_rate(df_train):.2%}')
```

```python
df_train_oversampled = oversampling(df_train)
print(f'New proportion of heart issues among samples {minority_class_rate(df_train_oversampled):.2%}')
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from sklearn.metrics import f1_score

from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators=500,        # Nombre d’arbres
    learning_rate=0.1,       # Taux d’apprentissage
    max_depth=10,             # Profondeur des arbres
    subsample=0.8,           # Sous-échantillonnage
    colsample_bytree=0.8,    # Sous-échantillonnage des colonnes
    eval_metric='logloss',   # Fonction de coût
    random_state=42,
    use_label_encoder=False
)

X_train = df_train_oversampled.drop(columns=['TARGET', 'ID'])
y_train = df_train_oversampled['TARGET']

model.fit(X_train, y_train)
```

```python
# Make predictions on the test set
y_pred = model.predict(X_train)

# Calculate the F1 score
f1 = f1_score(y_train, y_pred)

print(f"F1 Score: {f1:.2f}")
```
